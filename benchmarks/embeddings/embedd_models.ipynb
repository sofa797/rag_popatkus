```python
import os
import sys
import json
import time
import torch
import pandas as pd
import numpy as np
import gc
import matplotlib.pyplot as plt
import seaborn as sns
from sentence_transformers import SentenceTransformer, util

PROJECT_ROOT = "/content"
if PROJECT_ROOT not in sys.path:
    sys.path.append(PROJECT_ROOT)

DATA_PATH = "/content/parsed_pdf.json"
DATASET_PATH = "/content/test_dataset.json"

try:
    from chunker import Chunker
except ImportError as e:
    print(f"error: {e}")
    sys.exit(1)

def load_data():
    if not os.path.exists(DATA_PATH):
        raise FileNotFoundError(f"file is not found: {DATA_PATH}")
    with open(DATA_PATH, 'r', encoding='utf-8') as f:
        parsed_data = json.load(f)
    chunker_inst = Chunker()
    chunks = chunker_inst.chunk(parsed_data)
    return [c for c in chunks if str(c.get("text", "")).strip()]

def load_test_dataset():
    if not os.path.exists(DATASET_PATH):
        raise FileNotFoundError(f"dataset file is not found: {DATASET_PATH}")
    with open(DATASET_PATH, 'r', encoding='utf-8') as f:
        return json.load(f)

MODELS = {
    "MiniLM-L12": "sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2",
    "MPNet-Base": "sentence-transformers/paraphrase-multilingual-mpnet-base-v2",
    "E5-Large": "intfloat/multilingual-e5-large",
    "BGE-M3": "BAAI/bge-m3",
    "Distiluse": "sentence-transformers/distiluse-base-multilingual-cased-v2"
}

PREFIXES = {
    "E5-Large": {"query": "query: ", "doc": "passage: "},
    "GTE-Multi": {"query": "query: ", "doc": "passage: "}
}

def plot_all_metrics(df):
    sns.set_theme(style="whitegrid")
    plt.rcParams.update({
        'font.size': 11,
        'axes.labelsize': 12,
        'axes.titlesize': 13,
        'figure.titlesize': 16
    })

    output_dir = "/content/benchmark_plots"
    os.makedirs(output_dir, exist_ok=True)


    metrics_to_plot = ["Acc@1 (%)", "Acc@3 (%)", "Acc@5 (%)", "MRR"]
    df_melted = df.melt(id_vars="Model", value_vars=metrics_to_plot, var_name="Metric", value_name="Value")
    df_melted.loc[df_melted["Metric"] == "MRR", "Value"] *= 100
    df_melted.loc[df_melted["Metric"] == "MRR", "Metric"] = "MRR (x100)"

    plt.figure(figsize=(14, 7))
    ax = sns.barplot(x="Model", y="Value", hue="Metric", data=df_melted, palette="viridis")
    plt.title("Comparative analysis of the quality of search for embedding models", pad=20, weight='bold')
    plt.xlabel("Encoder model", labelpad=10)
    plt.ylabel("Metric value (%)", labelpad=10)
    plt.ylim(0, 110)
    plt.legend(title="Metrics", loc="upper left")
    for p in ax.patches:
        if p.get_height() > 0:
            ax.annotate(f"{p.get_height():.1f}%", (p.get_x() + p.get_width() / 2., p.get_height()),
                         ha='center', va='center', xytext=(0, 8), textcoords='offset points', fontsize=9, weight='semibold')
    plt.tight_layout()
    plt.savefig(f"{output_dir}/00_summary_quality.png", dpi=300)
    plt.show()

    fig, (ax2, ax3) = plt.subplots(1, 2, figsize=(16, 6))
    sns.barplot(x="Model", y="Avg Lat (ms)", data=df, ax=ax2, palette="magma")
    ax2.set_title("Average latency of one request (Avg Lat)", weight='bold', pad=15)
    ax2.set_xlabel("Model")
    ax2.set_ylabel("Time (ms)")
    for p in ax2.patches:
        ax2.annotate(f"{p.get_height():.2f} ms", (p.get_x() + p.get_width() / 2., p.get_height()), ha='center', va='center', xytext=(0, 5), textcoords='offset points', fontsize=10)

    sns.barplot(x="Model", y="Idx Time (s)", data=df, ax=ax3, palette="rocket")
    ax3.set_title("Indexing time for the entire corpus (Idx Time)", weight='bold', pad=15)
    ax3.set_xlabel("Model")
    ax3.set_ylabel("Time (s)")
    for p in ax3.patches:
        ax3.annotate(f"{p.get_height():.3f} s", (p.get_x() + p.get_width() / 2., p.get_height()), ha='center', va='center', xytext=(0, 5), textcoords='offset points', fontsize=10)
    plt.tight_layout()
    plt.savefig(f"{output_dir}/00_summary_latency.png", dpi=300)
    plt.show()


    all_metrics = [
        ("Acc@1 (%)", "Value of the metric (%)", "crest"),
        ("Acc@3 (%)", "Value of the metric (%)", "crest"),
        ("Acc@5 (%)", "Value of the metric (%)", "crest"),
        ("P@3 (%)", "Precision (%)", "flare"),
        ("P@5 (%)", "Precision (%)", "flare"),
        ("F1@3 (%)", "Value of F1-measure (%)", "mako"),
        ("F1@5 (%)", "Value of F1-measure (%)", "mako"),
        ("MRR", "MRR Ranking Index (0.0 - 1.0)", "viridis"),
        ("Avg Lat (ms)", "Delay time (ms)", "magma"),
        ("Idx Time (s)", "Case processing time (s)", "rocket")
    ]

    for metric_name, ylabel, palette in all_metrics:
        plt.figure(figsize=(9, 5))
        ax_m = sns.barplot(x="Model", y=metric_name, data=df, palette=palette)
        plt.title(f"Comparison of models by metrics {metric_name}", pad=15, weight='bold')
        plt.xlabel("Encoder model", labelpad=8)
        plt.ylabel(ylabel, labelpad=8)

        if "%" in metric_name:
            plt.ylim(0, 110)
        elif metric_name == "MRR":
            plt.ylim(0, 1.1)
        else:
            plt.ylim(0, df[metric_name].max() * 1.2)

        for p in ax_m.patches:
            val = p.get_height()
            if val > 0:
                fmt = f"{val:.2f}" if "%" not in metric_name else f"{val:.1f}%"
                ax_m.annotate(fmt, (p.get_x() + p.get_width() / 2., val),
                             ha='center', va='center', xytext=(0, 6),
                             textcoords='offset points', fontsize=10, weight='semibold')

        plt.tight_layout()
        safe_name = metric_name.replace("@", "_").replace(" (%)", "").replace(" (ms)", "_ms").replace(" (s)", "_s").lower()
        plt.savefig(f"{output_dir}/metric_{safe_name}.png", dpi=300)
        plt.show()

def run_benchmark():
    all_chunks = load_data()
    corpus_texts = [c["text"] for c in all_chunks if "text" in c]
    test_dataset = load_test_dataset()

    device = "cuda" if torch.cuda.is_available() else "cpu"
    print(f"\nthe number of chunks: {len(corpus_texts)} | queries: {len(test_dataset)} | device: {device.upper()}")
    results = []

    for name, path in MODELS.items():
        try:
            model = SentenceTransformer(path, device=device, trust_remote_code=True)
            doc_prefix = PREFIXES.get(name, {}).get("doc", "")
            prepared_corpus = [doc_prefix + t for t in corpus_texts]

            t0 = time.time()
            corpus_embeddings = model.encode(
                prepared_corpus, batch_size=32, show_progress_bar=False,
                convert_to_tensor=True, normalize_embeddings=True
            )
            idx_time = time.time() - t0

            metrics = {
                "acc_1": [], "acc_3": [], "acc_5": [],
                "p_3": [], "p_5": [], "f1_3": [], "f1_5": [], "mrr": []
            }

            query_prefix = PREFIXES.get(name, {}).get("query", "")
            prepared_queries = [query_prefix + item["query"] for item in test_dataset]

            t_latency_start = time.time()
            query_embeddings = model.encode(
                prepared_queries, batch_size=32, convert_to_tensor=True, normalize_embeddings=True
            )

            all_cos_scores = util.cos_sim(query_embeddings, corpus_embeddings)
            total_latency = time.time() - t_latency_start
            avg_latency = (total_latency / len(test_dataset)) * 1000

            for i, item in enumerate(test_dataset):
                cos_scores = all_cos_scores[i]
                top_k = torch.topk(cos_scores, k=min(5, len(corpus_texts)))
                found_texts = [corpus_texts[idx] for idx in top_k.indices.tolist()]

                for k in [1, 3, 5]:
                    slice_chunks = found_texts[:k]
                    is_found = item["target"] in slice_chunks

                    acc_val = 1.0 if is_found else 0.0
                    rec_val = 1.0 if is_found else 0.0
                    prec_val = (1.0 / k) if is_found else 0.0
                    f1_val = (2 * prec_val * rec_val) / (prec_val + rec_val) if is_found else 0.0

                    if k == 1:
                        metrics["acc_1"].append(acc_val)
                    else:
                        metrics[f"acc_{k}"].append(acc_val)
                        metrics[f"p_{k}"].append(prec_val)
                        metrics[f"f1_{k}"].append(f1_val)

                if item["target"] in found_texts:
                    rank = found_texts.index(item["target"]) + 1
                    metrics["mrr"].append(1.0 / rank)
                else:
                    metrics["mrr"].append(0.0)

            results.append({
                "Model": name,
                "Acc@1 (%)": round(np.mean(metrics["acc_1"]) * 100, 2),
                "Acc@3 (%)": round(np.mean(metrics["acc_3"]) * 100, 2),
                "Acc@5 (%)": round(np.mean(metrics["acc_5"]) * 100, 2),
                "P@3 (%)": round(np.mean(metrics["p_3"]) * 100, 2),
                "P@5 (%)": round(np.mean(metrics["p_5"]) * 100, 2),
                "F1@3 (%)": round(np.mean(metrics["f1_3"]) * 100, 2),
                "F1@5 (%)": round(np.mean(metrics["f1_5"]) * 100, 2),
                "MRR": round(np.mean(metrics["mrr"]), 3),
                "Avg Lat (ms)": round(avg_latency, 2),
                "Idx Time (s)": round(idx_time, 3)
            })

            del model, corpus_embeddings, query_embeddings, all_cos_scores
            if device == "cuda": torch.cuda.empty_cache()
            gc.collect()

        except Exception as e:
            print(f"testing error {name}: {e}")
            continue

    return pd.DataFrame(results)

if __name__ == "__main__":
    df = run_benchmark()
    if not df.empty:
        pd.set_option('display.max_columns', None)
        pd.set_option('display.width', 1000)
        print(df.to_string(index=False))
        plot_all_metrics(df)
```
